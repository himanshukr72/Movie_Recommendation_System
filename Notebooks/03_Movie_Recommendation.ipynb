## Parse TMDB JSON-like columns



```python

import json
import pandas as pd


def parse_tmdb_list(value):
    """Convert a JSON string into a Python list safely."""
    if isinstance(value, list):
        return value

    if not isinstance(value, str) or not value.strip():
        return []

    try:
        parsed = json.loads(value)
        return parsed if isinstance(parsed, list) else []
    except (json.JSONDecodeError, TypeError, ValueError):
        return []


def get_names(value, limit=None):
    """Extract unique names from a list of TMDB dictionaries."""
    items = parse_tmdb_list(value)
    names = []
    seen = set()

    for item in items:
        if not isinstance(item, dict):
            continue

        name = item.get("name")
        if isinstance(name, str) and name.strip():
            name = name.strip()

            if name not in seen:
                names.append(name)
                seen.add(name)

    return names if limit is None else names[:limit]


def get_director(value):
    """Extract the director's name from crew data."""
    crew = parse_tmdb_list(value)

    for member in crew:
        if (
            isinstance(member, dict)
            and member.get("job") == "Director"
        ):
            name = member.get("name")
            if isinstance(name, str) and name.strip():
                return name.strip()

    return ""


def get_writers(value):
    """Extract screenwriters and story writers from crew data."""
    crew = parse_tmdb_list(value)

    writer_jobs = {"Writer", "Screenplay", "Story"}
    writers = []
    seen = set()

    for member in crew:
        if not isinstance(member, dict):
            continue

        if member.get("job") in writer_jobs:
            name = member.get("name")

            if isinstance(name, str) and name.strip():
                name = name.strip()

                if name not in seen:
                    writers.append(name)
                    seen.add(name)

    return writers
```

## 6. Engineer content and metadata features

```python

# Drop rows that cannot be recommended because the identifier/title is missing.
movies = movies.dropna(subset=["movie_id", "title"]).copy()

# Handle missing overview text.
movies["overview"] = movies["overview"].fillna("").astype(str)

# Parse list-based TMDB metadata.
list_columns = {
    "genres": "genres_list",
    "keywords": "keywords_list",
    "production_companies": "production_companies_list",
    "production_countries": "production_countries_list",
}

for source_col, output_col in list_columns.items():
    if source_col in movies.columns:
        movies[output_col] = movies[source_col].apply(get_names)
    else:
        movies[output_col] = [[] for _ in range(len(movies))]

# Extract cast, director, and writers using the corrected JSON parsers.
movies["cast_top5"] = movies["cast"].apply(
    lambda value: get_names(value, limit=5)
)
movies["director"] = movies["crew"].apply(get_director)
movies["writers"] = movies["crew"].apply(get_writers)

# Extract release year.
movies["release_year"] = pd.to_datetime(
    movies.get("release_date", pd.Series("", index=movies.index)),
    errors="coerce"
).dt.year

# Convert numeric columns to appropriate types.
numeric_columns = [
    "runtime",
    "vote_average",
    "vote_count",
    "popularity",
    "budget",
    "revenue",
]

for column in numeric_columns:
    if column in movies.columns:
        movies[column] = pd.to_numeric(movies[column], errors="coerce")
    else:
        movies[column] = np.nan

# Clean language and poster metadata.
movies["original_language"] = movies.get(
    "original_language", pd.Series("", index=movies.index)
).fillna("").astype(str)

movies["poster_path"] = movies.get(
    "poster_path", pd.Series("", index=movies.index)
).fillna("").astype(str)

# Verify the results.
print("Rows after cleaning:", len(movies))
print(
    "Rows with cast:",
    movies["cast_top5"].apply(len).gt(0).sum()
)
print(
    "Rows with director:",
    movies["director"].str.strip().ne("").sum()
)
print(
    "Rows with writers:",
    movies["writers"].apply(len).gt(0).sum()
)

display(
    movies[
        [
            "movie_id",
            "title",
            "genres_list",
            "keywords_list",
            "cast_top5",
            "director",
            "writers",
            "production_companies_list",
            "production_countries_list",
            "original_language",
            "release_year",
        ]
    ].head(5)
)
```

## 9. BGE semantic embeddings for movie overviews

```python
import time
BGE_MODEL_NAME = "BAAI/bge-small-en-v1.5"
bge_model = SentenceTransformer(BGE_MODEL_NAME)

bge_inputs = feature_text["overview"].replace("", "No overview available").tolist()
start_time = time.time()

bge_embeddings = bge_model.encode(
    bge_inputs,
    batch_size=32,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True
).astype("float32")

print("Embedding shape:", bge_embeddings.shape)
print(f"Embedding time: {time.time() - start_time:.1f} seconds")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

feature_vectorizers = {}
feature_matrices = {}

structured_features = [
    "genres", "keywords", "cast", "director",
    "writers", "production", "country", "language"
]

for feature in structured_features:
    vectorizer = TfidfVectorizer(
        tokenizer=str.split,
        preprocessor=None,
        token_pattern=None,
        lowercase=True,
        norm="l2"
    )

    matrix = vectorizer.fit_transform(
        feature_text[feature].fillna("")
    )

    feature_vectorizers[feature] = vectorizer
    feature_matrices[feature] = matrix

    print(f"{feature:12s} -> {matrix.shape}")

print(
    "\nChristopher in vocabulary:",
    "christopher" in feature_vectorizers["director"].vocabulary_
)
print(
    "Nolan in vocabulary:",
    "nolan" in feature_vectorizers["director"].vocabulary_
)
```

```python
DEFAULT_WEIGHTS = {
    "bge": 0.40,
    "genres": 0.18,
    "keywords": 0.13,
    "cast": 0.10,
    "director": 0.06,
    "writers": 0.04,    
    "production": 0.04,
    "country": 0.03,
    "language": 0.02,
}

def normalize_weights(weights):
    clean_weights = {key: max(0.0, float(value)) for key, value in weights.items()}
    total = sum(clean_weights.values())
    if total <= 0:
        raise ValueError("At least one feature weight must be positive.")
    return {key: value / total for key, value in clean_weights.items()}

def weighted_similarity_scores(movie_index, weights=None):
    weights = normalize_weights(weights or DEFAULT_WEIGHTS)
    combined = np.zeros(len(model_df), dtype=np.float32)
    used_weight = 0.0

    if weights.get("bge", 0) > 0:
        bge_scores = bge_embeddings @ bge_embeddings[movie_index]
        combined += weights["bge"] * bge_scores
        used_weight += weights["bge"]

    for feature, matrix in feature_matrices.items():
        weight = weights.get(feature, 0.0)
        if weight > 0:
            scores = cosine_similarity(matrix[movie_index], matrix).ravel()
            combined += weight * scores.astype(np.float32)
            used_weight += weight

    if used_weight <= 0:
        raise ValueError("No active feature weights are available.")
    return combined / used_weight

def find_movie_index(movie_title):
    matches = model_df.index[
        model_df["title"].astype(str).str.casefold() == str(movie_title).casefold()
    ].tolist()
    if not matches:
        raise ValueError(f"Movie title not found: {movie_title}")
    return matches[0]

def recommend_weighted(movie_title, n=5, weights=None):
    movie_index = find_movie_index(movie_title)
    scores = weighted_similarity_scores(movie_index, weights)
    scores[movie_index] = -np.inf
    indices = np.argsort(scores)[::-1][:n]

    result = model_df.iloc[indices][["movie_id", "title", "genres_list", "release_year"]].copy()
    result["similarity_score"] = scores[indices]
    return result.reset_index(drop=True)
```

```python

import re
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity


def recommend_movies(
    movie_title=None,
    director=None,
    genres=None,
    writers=None,
    production=None,
    cast=None,
    keywords=None,
    country=None,
    language=None,
    overview=None,
    n=5,
    weights=None,
):

    if n < 1:
        raise ValueError("n must be at least 1.")

    # Map function arguments to the model's feature names.
    criteria = {
        "director": director,
        "genres": genres,
        "writers": writers,
        "production": production,
        "cast": cast,
        "keywords": keywords,
        "country": country,
        "language": language,
    }

    # Convert user inputs into text compatible with the TF-IDF models.
    def to_query_text(value):
        if value is None:
            return ""

        if isinstance(value, (list, tuple, set)):
            value = " ".join(str(item) for item in value)

        text = str(value).lower()
        text = re.sub(r"[^a-z0-9\s]", " ", text)
        return re.sub(r"\s+", " ", text).strip()

    # Find the starting movie, if provided.
    seed_index = None

    if movie_title is not None:
        seed_index = find_movie_index(movie_title)

    # Build a query for each structured feature.
    query_features = {}

    if seed_index is not None:
        # Start with the selected movie's existing feature values.
        for feature in feature_matrices:
            query_features[feature] = to_query_text(
                feature_text.iloc[seed_index][feature]
            )

    # Explicit criteria override the starting movie's feature values.
    for feature, value in criteria.items():
        if value is not None:
            query_features[feature] = to_query_text(value)

    # Prepare weights. Custom weights override the defaults.
    weight_config = DEFAULT_WEIGHTS.copy()

    if weights is not None:
        weight_config.update(weights)

    weight_config = {
        key: max(0.0, float(value))
        for key, value in weight_config.items()
    }

    combined_scores = np.zeros(len(model_df), dtype=np.float32)
    total_weight = 0.0

    # Compare each supplied feature against all movies.
    for feature, query_text in query_features.items():

        if not query_text or feature not in feature_matrices:
            continue

        weight = weight_config.get(feature, 0.0)

        if weight <= 0:
            continue

        query_text = query_text.replace(" ", "_") if feature == "director" else query_text

        query_vector = feature_vectorizers[feature].transform(
            [query_text]
        )

        # Skip queries containing no terms known to the vectorizer.
        if query_vector.nnz == 0:
            continue

        scores = cosine_similarity(
            query_vector,
            feature_matrices[feature]
        ).ravel()

        combined_scores += weight * scores.astype(np.float32)
        total_weight += weight

    # Add BGE semantic similarity when a movie or overview is supplied.
    if overview is not None:
        bge_query = str(overview).strip()
        use_bge = bool(bge_query)

        if use_bge:
            query_embedding = bge_model.encode(
                [bge_query],
                convert_to_numpy=True,
                normalize_embeddings=True,
            )[0]

    elif seed_index is not None:
        query_embedding = bge_embeddings[seed_index]
        use_bge = True

    else:
        use_bge = False

    bge_weight = weight_config.get("bge", 0.0)

    if use_bge and bge_weight > 0:
        bge_scores = bge_embeddings @ query_embedding
        combined_scores += bge_weight * bge_scores.astype(np.float32)
        total_weight += bge_weight

    if total_weight <= 0:
        raise ValueError(
            "No usable query features found. Try a movie title "
            "or provide a valid director, genre, cast, or other criterion."
        )

    # Normalize the score by the weights actually used.
    combined_scores /= total_weight

    # Do not recommend the starting movie itself.
    if seed_index is not None:
        combined_scores[seed_index] = -np.inf

    # Select the highest-scoring recommendations.
    indices = np.argsort(combined_scores)[::-1][:n]

    result = model_df.iloc[indices][
        ["movie_id", "title", "genres_list", "release_year"]
    ].copy()

    result["similarity_score"] = combined_scores[indices]

    return result.reset_index(drop=True)
```
