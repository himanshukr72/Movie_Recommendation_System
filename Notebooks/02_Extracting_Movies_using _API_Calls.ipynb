```python

import os
import json
import time
import warnings
import requests
import pandas as pd

from pathlib import Path
from dotenv import load_dotenv

warnings.filterwarnings("default")

# Retry configuration
WAIT_SECONDS = 10
MAX_RETRIES = 6
REQUEST_TIMEOUT = 30
REQUEST_INTERVAL = 0.12

# Project root: notebook is assumed to be in Notebooks/
PROJECT_ROOT = Path.cwd().parent

# Load API key
ENV_FILE = PROJECT_ROOT / ".env"
load_dotenv(ENV_FILE, override=True)

API_KEY = os.getenv("TMDB_API_KEY")

if not API_KEY:
    raise ValueError(
        f"TMDB_API_KEY not found. Check {ENV_FILE}"
    )

BASE_URL = "https://api.themoviedb.org/3"
TARGET_MOVIES = 10_000

# Output directory
DATA_DIR = PROJECT_ROOT / "Data" / "tmdb"
DATA_DIR.mkdir(parents=True, exist_ok=True)

# Checkpoints and output files
DISCOVERY_FILE = DATA_DIR / "discovered_movies.json"
MOVIES_JSONL = DATA_DIR / "movies_checkpoint.jsonl"
CREDITS_JSONL = DATA_DIR / "credits_checkpoint.jsonl"
RAW_FILE = DATA_DIR / "raw_movie_details.jsonl"
FAILED_FILE = DATA_DIR / "failed_movie_ids.json"

MOVIES_FILE = DATA_DIR / "tmdb_movies.csv"
CREDITS_FILE = DATA_DIR / "tmdb_credits.csv"

session = requests.Session()

print("API key loaded:", bool(API_KEY))
print("Environment file exists:", ENV_FILE.is_file())
print("Data directory:", DATA_DIR)
print("Retry wait:", WAIT_SECONDS, "seconds")
print("Target movies:", TARGET_MOVIES)
```

```python

def tmdb_get(endpoint, params=None):
    """Request TMDB data with retries and warning messages."""

    params = dict(params or {})
    params["api_key"] = API_KEY

    url = f"{BASE_URL}/{endpoint.lstrip('/')}"

    for attempt in range(1, MAX_RETRIES + 1):
        try:
            response = session.get(
                url,
                params=params,
                timeout=REQUEST_TIMEOUT
            )

            if response.status_code == 429 or response.status_code >= 500:
                message = (
                    f"HTTP {response.status_code} for {endpoint}. "
                    f"Attempt {attempt}/{MAX_RETRIES}."
                )

                if attempt == MAX_RETRIES:
                    response.raise_for_status()

                warnings.warn(
                    message + f" Waiting {WAIT_SECONDS} seconds.",
                    RuntimeWarning
                )
                time.sleep(WAIT_SECONDS)
                continue

            # Authentication errors and other non-retryable
            # HTTP errors are raised immediately.
            response.raise_for_status()
            return response.json()

        except requests.RequestException as error:
            if attempt == MAX_RETRIES:
                warnings.warn(
                    f"All attempts failed for {endpoint}: {error}",
                    RuntimeWarning
                )
                raise

            warnings.warn(
                f"Request failed for {endpoint}: {error}. "
                f"Waiting {WAIT_SECONDS} seconds before retry "
                f"{attempt + 1}/{MAX_RETRIES}.",
                RuntimeWarning
            )
            time.sleep(WAIT_SECONDS)

    raise RuntimeError(f"TMDB request failed: {endpoint}")
```

```python

def save_json_atomic(path, obj):
    """Write a JSON file through a temporary file."""

    temp_path = path.with_suffix(path.suffix + ".tmp")

    with temp_path.open("w", encoding="utf-8") as file:
        json.dump(obj, file, ensure_ascii=False)

    temp_path.replace(path)


def read_jsonl(path):
    """Read JSON Lines records and warn about malformed lines."""

    records = []

    if not path.exists():
        return records

    with path.open("r", encoding="utf-8") as file:
        for line_number, line in enumerate(file, start=1):
            try:
                records.append(json.loads(line))
            except json.JSONDecodeError:
                warnings.warn(
                    f"Invalid JSON in {path}, line {line_number}.",
                    RuntimeWarning
                )

    return records


def append_jsonl(path, record):
    """Append one record immediately to a checkpoint file."""

    with path.open("a", encoding="utf-8") as file:
        file.write(json.dumps(record, ensure_ascii=False) + "\n")
        file.flush()


def as_json(value):
    """Serialize nested TMDB data into JSON strings for CSV."""

    return json.dumps(
        value if value is not None else [],
        ensure_ascii=False
    )
```

```python

# Load discovery checkpoint
if DISCOVERY_FILE.exists():
    with DISCOVERY_FILE.open("r", encoding="utf-8") as file:
        saved_movies = json.load(file)

    discovered = {
        int(movie["id"]): movie
        for movie in saved_movies
    }
else:
    discovered = {}


sort_orders = [
    "popularity.desc",
    "vote_count.desc",
    "primary_release_date.desc"
]

for sort_order in sort_orders:

    if len(discovered) >= TARGET_MOVIES:
        break

    # Determine where this query left off.
    pages_done = [
        int(movie.get("_discovery_page", 0))
        for movie in discovered.values()
        if movie.get("_discovery_sort") == sort_order
    ]

    start_page = max(pages_done, default=0) + 1

    for page in range(start_page, 501):

        if len(discovered) >= TARGET_MOVIES:
            break

        try:
            result = tmdb_get(
                "discover/movie",
                params={
                    "language": "en-US",
                    "include_adult": False,
                    "include_video": False,
                    "sort_by": sort_order,
                    "page": page
                }
            )

        except requests.RequestException as error:
            warnings.warn(
                f"Discovery failed for {sort_order}, page {page}: "
                f"{error}. Rerun this cell to resume.",
                RuntimeWarning
            )
            break

        for movie in result.get("results", []):
            movie_id = int(movie["id"])

            if movie_id not in discovered:
                discovered[movie_id] = {
                    **movie,
                    "_discovery_page": page,
                    "_discovery_sort": sort_order
                }

            if len(discovered) >= TARGET_MOVIES:
                break

        save_json_atomic(
            DISCOVERY_FILE,
            list(discovered.values())
        )

        print(
            f"{sort_order} | page {page} | "
            f"{len(discovered):,}/{TARGET_MOVIES:,} unique movies"
        )

        if page >= result.get("total_pages", 0):
            break

        time.sleep(REQUEST_INTERVAL)


movie_ids = list(discovered.keys())[:TARGET_MOVIES]

print("\nTotal unique movie IDs:", len(movie_ids))

if len(movie_ids) < TARGET_MOVIES:
    warnings.warn(
        f"Only {len(movie_ids):,} unique IDs were discovered. "
        "Run additional discovery queries or review the API results.",
        RuntimeWarning
    )
```

```python

MOVIE_COLUMNS = [
    "budget",
    "genres",
    "homepage",
    "id",
    "keywords",
    "original_language",
    "original_title",
    "overview",
    "popularity",
    "production_companies",
    "production_countries",
    "release_date",
    "revenue",
    "runtime",
    "spoken_languages",
    "status",
    "tagline",
    "title",
    "vote_average",
    "vote_count"
]

CREDIT_COLUMNS = [
    "movie_id",
    "title",
    "cast",
    "crew"
]

print("Movies columns:", len(MOVIE_COLUMNS))
print("Credits columns:", len(CREDIT_COLUMNS))
```

```python

def collect_movie(movie_id):
    """Fetch one movie and construct both table rows."""

    movie = tmdb_get(
        f"movie/{movie_id}",
        params={
            "language": "en-US",
            "append_to_response": "keywords,credits"
        }
    )

    if not movie.get("id"):
        raise ValueError(
            f"No valid movie details returned for ID {movie_id}"
        )

    keyword_data = movie.get("keywords") or {}
    credits_data = movie.get("credits") or {}

    movie_keywords = keyword_data.get("keywords", [])
    cast = credits_data.get("cast", [])
    crew = credits_data.get("crew", [])

    # Warnings identify incomplete metadata without stopping collection.
    if not movie.get("overview"):
        warnings.warn(
            f"Missing overview for movie ID {movie_id}",
            RuntimeWarning
        )

    if not movie_keywords:
        warnings.warn(
            f"Missing keywords for movie ID {movie_id}",
            RuntimeWarning
        )

    if not cast:
        warnings.warn(
            f"Missing cast for movie ID {movie_id}",
            RuntimeWarning
        )

    movie_row = {
        "budget": movie.get("budget"),
        "genres": as_json(movie.get("genres", [])),
        "homepage": movie.get("homepage"),
        "id": movie.get("id"),
        "keywords": as_json(movie_keywords),
        "original_language": movie.get("original_language"),
        "original_title": movie.get("original_title"),
        "overview": movie.get("overview"),
        "popularity": movie.get("popularity"),
        "production_companies": as_json(
            movie.get("production_companies", [])
        ),
        "production_countries": as_json(
            movie.get("production_countries", [])
        ),
        "release_date": movie.get("release_date"),
        "revenue": movie.get("revenue"),
        "runtime": movie.get("runtime"),
        "spoken_languages": as_json(
            movie.get("spoken_languages", [])
        ),
        "status": movie.get("status"),
        "tagline": movie.get("tagline"),
        "title": movie.get("title"),
        "vote_average": movie.get("vote_average"),
        "vote_count": movie.get("vote_count")
    }

    credit_row = {
        "movie_id": movie.get("id"),
        "title": movie.get("title"),
        "cast": as_json(cast),
        "crew": as_json(crew)
    }

    return movie_row, credit_row, movie
```

```python

def export_checkpoints():
    movie_rows = read_jsonl(MOVIES_JSONL)
    credit_rows = read_jsonl(CREDITS_JSONL)

    movies_df = pd.DataFrame(
        movie_rows,
        columns=MOVIE_COLUMNS
    )

    credits_df = pd.DataFrame(
        credit_rows,
        columns=CREDIT_COLUMNS
    )

    if not movies_df.empty:
        movies_df = movies_df.drop_duplicates(
            subset=["id"],
            keep="last"
        )

    if not credits_df.empty:
        credits_df = credits_df.drop_duplicates(
            subset=["movie_id"],
            keep="last"
        )

    movies_df.to_csv(MOVIES_FILE, index=False)
    credits_df.to_csv(CREDITS_FILE, index=False)

    return movies_df, credits_df
```

```python

# Load completed IDs from both tables.
completed_movies = {
    int(row["id"])
    for row in read_jsonl(MOVIES_JSONL)
    if row.get("id") is not None
}

completed_credits = {
    int(row["movie_id"])
    for row in read_jsonl(CREDITS_JSONL)
    if row.get("movie_id") is not None
}

completed_ids = completed_movies & completed_credits

# Warn if the checkpoint files are inconsistent.
if completed_movies != completed_credits:
    warnings.warn(
        f"Checkpoint mismatch: {len(completed_movies)} movie records "
        f"and {len(completed_credits)} credit records. "
        "Records missing from either table will be retried.",
        RuntimeWarning
    )

# Load previously failed IDs.
if FAILED_FILE.exists():
    with FAILED_FILE.open("r", encoding="utf-8") as file:
        failed_ids = set(map(int, json.load(file)))
else:
    failed_ids = set()

if failed_ids:
    warnings.warn(
        f"{len(failed_ids)} movie IDs previously failed. "
        "They will be retried during this collection pass.",
        RuntimeWarning
    )

successful_since_save = 0
total_movies = len(movie_ids)

print(f"Target movie IDs: {total_movies:,}")
print(f"Already completed: {len(completed_ids):,}")
print(f"Remaining: {total_movies - len(completed_ids):,}")

for index, movie_id in enumerate(movie_ids, start=1):

    if movie_id in completed_ids:
        continue

    try:
        movie_row, credit_row, raw = collect_movie(movie_id)

        # Warn if the API returns an unexpected ID.
        if int(movie_row["id"]) != int(movie_id):
            warnings.warn(
                f"Requested ID {movie_id}, but received "
                f"ID {movie_row['id']}. Skipping this record.",
                RuntimeWarning
            )
            failed_ids.add(movie_id)
            continue

        # Save movie and credit records.
        append_jsonl(MOVIES_JSONL, movie_row)
        append_jsonl(CREDITS_JSONL, credit_row)

        # Preserve the complete response for future reprocessing.
        append_jsonl(
            RAW_FILE,
            {
                "id": movie_id,
                "data": raw
            }
        )

        completed_ids.add(movie_id)
        failed_ids.discard(movie_id)
        successful_since_save += 1

        print(
            f"[{index}/{total_movies}] "
            f"Completed: {len(completed_ids):,}/{total_movies:,} "
            f"| {movie_row['title']}"
        )

    except (
        requests.RequestException,
        ValueError,
        KeyError
    ) as error:

        failed_ids.add(movie_id)

        warnings.warn(
            f"Failed to collect movie ID {movie_id}: {error}. "
            "The ID has been recorded for a future retry.",
            RuntimeWarning
        )

    except Exception as error:
        # Unexpected errors should be visible rather than silently lost.
        failed_ids.add(movie_id)

        warnings.warn(
            f"Unexpected error for movie ID {movie_id}: "
            f"{type(error).__name__}: {error}",
            RuntimeWarning
        )

    # Save a checkpoint every 50 successful movies.
    if successful_since_save >= 50:
        movies, credits = export_checkpoints()

        save_json_atomic(
            FAILED_FILE,
            sorted(failed_ids)
        )

        print(
            f"Checkpoint saved: {len(completed_ids):,} completed, "
            f"{len(failed_ids):,} failed."
        )

        successful_since_save = 0

    # Persist failed IDs periodically, even if there are few successes.
    if index % 10 == 0:
        save_json_atomic(
            FAILED_FILE,
            sorted(failed_ids)
        )

    time.sleep(REQUEST_INTERVAL)

# Final checkpoint.
movies, credits = export_checkpoints()

save_json_atomic(
    FAILED_FILE,
    sorted(failed_ids)
)

print("\nCollection pass finished.")
print("Completed movie records:", movies["id"].nunique())
print("Completed credit records:", credits["movie_id"].nunique())
print("Failed IDs remaining:", len(failed_ids))
print("Movies file:", MOVIES_FILE)
print("Credits file:", CREDITS_FILE)

# Final validation warnings.
if movies["id"].nunique() < TARGET_MOVIES:
    warnings.warn(
        f"Only {movies['id'].nunique():,} unique movies collected; "
        f"target is {TARGET_MOVIES:,}.",
        RuntimeWarning
    )

if len(failed_ids) > 0:
    warnings.warn(
        f"{len(failed_ids)} movie IDs remain unsuccessful. "
        "Run the failed-ID retry cell to try again.",
        RuntimeWarning
    )

if movies["id"].duplicated().any():
    warnings.warn(
        "Duplicate movie IDs detected in the final movies table.",
        RuntimeWarning
    )

if credits["movie_id"].duplicated().any():
    warnings.warn(
        "Duplicate movie IDs detected in the final credits table.",
        RuntimeWarning
    )
```

```python

if FAILED_FILE.exists():
    with FAILED_FILE.open("r", encoding="utf-8") as file:
        failed_ids = set(map(int, json.load(file)))
else:
    failed_ids = set()

print("Failed movies to retry:", len(failed_ids))

for movie_id in sorted(failed_ids):

    try:
        movie_row, credit_row, raw = collect_movie(movie_id)

        # Remove any earlier record for this ID before appending.
        # This avoids duplicate checkpoint records on repeated retries.
        movie_records = [
            row for row in read_jsonl(MOVIES_JSONL)
            if int(row.get("id", -1)) != movie_id
        ]

        credit_records = [
            row for row in read_jsonl(CREDITS_JSONL)
            if int(row.get("movie_id", -1)) != movie_id
        ]

        # Rewrite checkpoints with the updated records.
        with MOVIES_JSONL.open("w", encoding="utf-8") as file:
            for row in movie_records + [movie_row]:
                file.write(json.dumps(row, ensure_ascii=False) + "\n")

        with CREDITS_JSONL.open("w", encoding="utf-8") as file:
            for row in credit_records + [credit_row]:
                file.write(json.dumps(row, ensure_ascii=False) + "\n")

        append_jsonl(
            RAW_FILE,
            {"id": movie_id, "data": raw}
        )

        failed_ids.discard(movie_id)
        print(f"Retry succeeded: {movie_row['title']}")

    except (
        requests.RequestException,
        ValueError,
        KeyError
    ) as error:
        warnings.warn(
            f"Retry failed for movie ID {movie_id}: {error}",
            RuntimeWarning
        )

    save_json_atomic(FAILED_FILE, sorted(failed_ids))
    time.sleep(REQUEST_INTERVAL)

movies, credits = export_checkpoints()

print("Remaining failed IDs:", len(failed_ids))
print("Movies rows:", len(movies))
print("Credits rows:", len(credits))
```
