```python
import requests
import pandas as pd
import time
import warnings
API_KEY= "f0d50aac24ed322d48d305c472e1cedb"

movie_df = pd.DataFrame()

for i in range(1, 511):

    url = "https://api.themoviedb.org/3/movie/popular"

    params = {
        "api_key": API_KEY,
        "language": "en-US",
        "page": i
    }

    while True:

        try:
            response = requests.get(
                url,
                params=params,
                headers={"accept": "application/json"},
                timeout=30
            )

            if response.status_code == 200:
                
                data = response.json()

                df = pd.DataFrame(data["results"])
                

                movie_df = pd.concat(
                    [movie_df, df],
                    ignore_index=True
                )

                print(f"Page {i} completed")
                break

            else:
                print(f"Page {i} returned status: {response.status_code}")
                print(response.text)
                break

        except requests.exceptions.ConnectionError:
            print(f"Connection reset on page {i}. Waiting 10 seconds...")
            time.sleep(10)

        except requests.exceptions.Timeout:
            print(f"Timeout on page {i}. Waiting 10 seconds...")
            time.sleep(10)

    # Wait before requesting the next page
    time.sleep(3)

print("Total movies:", len(movie_df))
```
